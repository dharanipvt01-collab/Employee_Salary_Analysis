# Employee Salary Analysis

## Internship Project - Veda Technology

This project analyzes employee salary data using Python.

The main objectives are to calculate:

- Total salary
- Average salary
- Highest salary
- Lowest salary
- Highest-paid employee

The project also uses Pandas for data processing and Matplotlib
for data visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Employee": [
        "Arun",
        "Priya",
        "Karthik",
        "Divya",
        "Rahul",
        "Sneha",
        "Vijay",
        "Anjali",
        "Surya",
        "Keerthana"
    ],
    "Salary": [
        25000,
        32000,
        28000,
        45000,
        38000,
        30000,
        52000,
        35000,
        41000,
        29000
    ]
}

df = pd.DataFrame(data)

print("Employee Salary Dataset")
df
```

```python
print("Number of Employees:", len(df))
print("Columns:", list(df.columns))
```

```python
salaries = df["Salary"].tolist()

print("Salary Values:")
print(salaries)
```

```python
total_salary = sum(salaries)

print("Total Salary:", total_salary)
```

```python
average_salary = total_salary / len(salaries)

print("Average Salary:", average_salary)
```

```python
highest_salary = max(salaries)

print("Highest Salary:", highest_salary)
```

```python
lowest_salary = min(salaries)

print("Lowest Salary:", lowest_salary)
```

```python
highest_paid_employee = df.loc[df["Salary"].idxmax()]

print("Highest-Paid Employee:", highest_paid_employee["Employee"])
print("Salary:", highest_paid_employee["Salary"])
```

```python
pandas_total = df["Salary"].sum()
pandas_average = df["Salary"].mean()
pandas_highest = df["Salary"].max()
pandas_lowest = df["Salary"].min()

print("Salary Analysis using Pandas")
print("----------------------------")
print("Total Salary:", pandas_total)
print("Average Salary:", pandas_average)
print("Highest Salary:", pandas_highest)
print("Lowest Salary:", pandas_lowest)
```

```python
summary = pd.DataFrame({
    "Statistic": [
        "Total Salary",
        "Average Salary",
        "Highest Salary",
        "Lowest Salary"
    ],
    "Value": [
        total_salary,
        average_salary,
        highest_salary,
        lowest_salary
    ]
})

summary
```

```python
print("Salary Range Analysis")
print("---------------------")
print("Lowest Salary :", lowest_salary)
print("Highest Salary:", highest_salary)
print("Difference    :", highest_salary - lowest_salary)
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Employee"], df["Salary"])

plt.xlabel("Employee")
plt.ylabel("Salary")
plt.title("Employee Salary Analysis")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Employee"], df["Salary"])
plt.axhline(
    average_salary,
    linestyle="--",
    label=f"Average Salary = {average_salary:.0f}"
)

plt.xlabel("Employee")
plt.ylabel("Salary")
plt.title("Employee Salary vs Average Salary")

plt.xticks(rotation=45)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
print("=" * 40)
print("       EMPLOYEE SALARY ANALYSIS")
print("=" * 40)

print(f"Number of Employees : {len(df)}")
print(f"Total Salary        : ₹{total_salary:,.2f}")
print(f"Average Salary      : ₹{average_salary:,.2f}")
print(f"Highest Salary      : ₹{highest_salary:,.2f}")
print(f"Lowest Salary       : ₹{lowest_salary:,.2f}")
print(
    f"Highest Paid        : {highest_paid_employee['Employee']}"
)

print("=" * 40)
```
